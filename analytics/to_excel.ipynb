```python
import pandas as pd
from connection import ConnectionManager
```

```python
query = f"""
        SELECT
            *
        FROM
            TICK_DATA
        WHERE
            INSTRUMENT_ID = 2504
            AND CREATED_AT <= '2026-03-05 15:30:00'
            AND CREATED_AT >= '2026-03-05 09:15:00'
        ORDER BY
            TIMESTAMP;
    """
manager = ConnectionManager()
conn = manager.get_connection()
with conn.cursor() as cur:
    cur.execute(query)
    res = cur.fetchall()
res
```

```python
df = pd.DataFrame(res)
df.to_excel('2026-03-05.xlsx', index=False)
```
